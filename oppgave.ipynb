**Sympy** er et bibliotek som lar Python regne med bokstaver, akkurat som CAS i GeoGebra. Forskjellen fra vanlig Python er at sympy gir **eksakte svar**: $\sqrt{8}$ blir $2\sqrt{2}$, ikke `2.8284271247461903`, og $\frac{1}{3}$ forblir en brøk.

Hvorfor sympy i stedet for GeoGebra CAS?

- **Pålitelig.** Sympy løser kompliserte likninger og likningssett uten å «henge seg», og du ser nøyaktig hvilke kommandoer som ga svaret.
- **Alt på ett sted.** Utregning, grafer og forklaring ligger i samme notebook.
- **Kan kombineres med programmering.** Du kan løse hundre likninger i en løkke, eller lage dine egne funksjoner.

## Eksakte svar og desimaltall

Vanlig Python regner med desimaltall. Sympy regner eksakt så lenge du gir den eksakte tall.

| Skriv | Betyr | Gir |
| --- | --- | --- |
| `sp.Rational(1, 3)` | brøken $\frac{1}{3}$ | `1/3` |
| `sp.sqrt(8)` | $\sqrt{8}$ | `2*sqrt(2)` |
| `sp.pi` | $\pi$ | `pi` |
| `sp.N(uttrykk)` | desimalverdien (som ≈-knappen i GeoGebra) | `2.82842712474619` |

**Pass på:** `1/3` i vanlig Python blir straks `0.333…`. Skal du ha en eksakt brøk, bruk `sp.Rational(1, 3)`.

```python
print(1/3)                     # vanlig Python: desimaltall
print(sp.Rational(1, 3))       # sympy: eksakt brøk
print(sp.sqrt(8))              # forenkles automatisk
print(sp.N(sp.sqrt(8)))        # desimalverdi
print(sp.N(sp.pi, 30))         # pi med 30 siffer
```

## Omforme uttrykk

| Sympy | Gjør | GeoGebra CAS |
| --- | --- | --- |
| `sp.expand(u)` | multipliserer ut parenteser | `Utvid(u)` |
| `sp.factor(u)` | faktoriserer | `Faktoriser(u)` |
| `sp.simplify(u)` | forenkler så godt den kan | `Forenkle(u)` |
| `u.subs(x, 2)` | setter inn $x = 2$ | bytte ut $x$ med en verdi |

```python
print(sp.expand((x + 3)*(x - 1)))
print(sp.factor(x**2 - 9))
print(sp.simplify((x**2 + 2*x + 1)/(x + 1)))
print((x**2 + 1).subs(x, 3))
```

## Løse likninger

En likning skrives med `sp.Eq(venstre side, høyre side)`. Deretter løser vi med `sp.solve(likning, x)`, som svarer til `Løs` i GeoGebra. Svaret er en **liste** med løsninger.

Er høyre side 0, kan du sløyfe `sp.Eq` og bare skrive uttrykket.

```python
likning = sp.Eq(x**2 + 2*x, 8)
print(sp.solve(likning, x))

print(sp.solve(x**2 - 2, x))      # samme som x**2 - 2 = 0
```

## Likningssett

Et likningssett er en liste med likninger. Vi sier også hvilke ukjente vi skal finne. Svaret blir en **dict**, der vi slår opp verdien til hver ukjent.

$$\begin{aligned} x + y &= 10 \\ x - y &= 4 \end{aligned}$$

```python
losning = sp.solve([sp.Eq(x + y, 10), sp.Eq(x - y, 4)], [x, y])
print(losning)
print("x =", losning[x], "og y =", losning[y])
```

## Oversikt: fra GeoGebra CAS til sympy

| Du vil | GeoGebra CAS | Sympy |
| --- | --- | --- |
| Løse en likning | `Løs(x^2 = 4, x)` | `sp.solve(sp.Eq(x**2, 4), x)` |
| Løse numerisk | `NLøs(...)` | `sp.nsolve(uttrykk, x, startverdi)` |
| Løse likningssett | `Løs({likn1, likn2}, {x, y})` | `sp.solve([likn1, likn2], [x, y])` |
| Faktorisere | `Faktoriser(...)` | `sp.factor(...)` |
| Multiplisere ut | `Utvid(...)` | `sp.expand(...)` |
| Forenkle | `Forenkle(...)` | `sp.simplify(...)` |
| Derivere | `Derivert(f(x), x)` | `sp.diff(f, x)` |
| Desimalsvar | ≈-knappen | `sp.N(...)` |

**Husk forskjellene fra GeoGebra:** gangetegn må alltid skrives (`2*x`, ikke `2x`), og potens skrives `**` (`x**2`, ikke `x^2`).

```python
# Kjør denne cellen for å sjekke svarene dine. Du trenger ikke forstå koden.
def naer(a, b, rel=1e-3):
    """Sant hvis tallet a er innenfor rel (relativt) av b."""
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if b == 0:
        return abs(a) < 1e-9
    return abs(a - b) <= rel * abs(b)


def S(tekst):
    """Gjør tekst om til et sympy-uttrykk, f.eks. S("x**2 - 1")."""
    import sympy
    return sympy.sympify(tekst, locals={n: sympy.Symbol(n) for n in ("N", "O", "Q", "S")})


def lik(a, b):
    """Sant hvis a og b er matematisk like.

    Lister og mengder sammenlignes uten hensyn til rekkefølge, tupler (koordinater) i rekkefølge.
    """
    import sympy
    if a is None:
        return False
    if isinstance(b, tuple):
        if not isinstance(a, (tuple, list, sympy.Tuple)) or len(a) != len(b):
            return False
        return all(lik(x, y) for x, y in zip(a, b))
    if isinstance(b, (list, set)):
        if isinstance(a, sympy.FiniteSet):
            a = list(a)
        if not isinstance(a, (list, tuple, set)) or len(a) != len(b):
            return False
        rest = list(a)
        for y in b:
            for i, x in enumerate(rest):
                if lik(x, y):
                    del rest[i]
                    break
            else:
                return False
        return True
    if isinstance(b, dict):
        if not isinstance(a, dict) or len(a) != len(b):
            return False
        a = {str(k): v for k, v in a.items()}
        return all(str(k) in a and lik(a[str(k)], v) for k, v in b.items())
    try:
        a, b = sympy.sympify(a), sympy.sympify(b)
    except (sympy.SympifyError, TypeError):
        return False
    vanlige = lambda u: u.subs({s: sympy.Symbol(s.name) for s in u.free_symbols})
    a, b = vanlige(a), vanlige(b)
    if isinstance(a, sympy.Set) or isinstance(b, sympy.Set):
        return a == b
    if getattr(b, "is_Relational", False):
        return a == b
    try:
        return sympy.simplify(a - b) == 0
    except TypeError:
        return a == b

SJEKKER = [
    ('1a', 'svar_1a skal være (2x - 3)² multiplisert ut', "svar_1a == S('4*x**2 - 12*x + 9')"),
    ('1b', 'svar_1b skal være x² - 5x + 6 faktorisert', "svar_1b == S('(x - 2)*(x - 3)')"),
    ('1c', 'svar_1c skal være (x² - 9)/(x + 3) forenklet', "svar_1c == S('x - 3')"),
    ('1d', 'svar_1d skal være den eksakte verdien (en brøk) av x³ - 2x for x = 1/2', "svar_1d == S('-7/8')"),
    ('2a', 'losning_2a skal være løsningene av x² - 2x - 15 = 0', 'lik(losning_2a, [-3, 5])'),
    ('2b', 'losning_2b skal være løsningene av 3/x + 1 = x/2', "lik(losning_2b, [S('1 - sqrt(7)'), S('1 + sqrt(7)')])"),
    ('2c', 'losning_2c skal være løsningene av x³ - 6x² + 11x - 6 = 0', 'lik(losning_2c, [1, 2, 3])'),
    ('3', 'losning_3 skal være løsningen av likningssettet i oppgave 3', "lik(losning_3, {'x': 3, 'y': 2})"),
    ('4_to', 'los_andregrad(1, -5, 6) skal gi [2, 3]', 'lik(los_andregrad(1, -5, 6), [2, 3])'),
    ('4_en', 'los_andregrad(1, -4, 4) skal gi [2]', 'lik(los_andregrad(1, -4, 4), [2])'),
    ('4_ingen', 'los_andregrad(1, 2, 5) skal gi [] (ingen reelle løsninger)', 'los_andregrad(1, 2, 5) == []'),
    ('4_eksakt', 'los_andregrad(2, 0, -1) skal gi de eksakte løsningene ±√2/2', "lik(los_andregrad(2, 0, -1), [S('-sqrt(2)/2'), S('sqrt(2)/2')])"),
    ('5', 'voksen og barn skal være prisene på billettene', 'lik(voksen, 160) and lik(barn, 100)'),
    ('6', 'losning_6 skal være skjæringspunktene mellom sirkelen og linjen', 'lik(losning_6, [(3, 4), (-4, -3)])'),
    ('7', 'k_verdier skal være verdiene av k som gir nøyaktig én løsning', 'lik(k_verdier, [-6, 6])'),
]


def test_alt():
    feil = 0
    for _, beskrivelse, uttrykk in SJEKKER:
        try:
            ok = eval(uttrykk)
        except Exception as e:
            ok = False
            beskrivelse += f" (krasjet: {type(e).__name__}: {e})"
        if not ok:
            print("❌", beskrivelse)
            feil += 1
    if feil == 0:
        print(f"✅ Alle testene bestått! ({len(SJEKKER)} av {len(SJEKKER)})")
    else:
        print(f"\n{len(SJEKKER) - feil} av {len(SJEKKER)} tester bestått.")


test_alt()
```
